```python
import sys, platform, importlib.util
from importlib.metadata import version, PackageNotFoundError

print("Python:", sys.version.split()[0])
print("OS:", platform.platform())

# Paket från requirements.txt (importnamn -> paketnamn)
paket = {"numpy": "numpy", "scipy": "scipy", "matplotlib": "matplotlib",
         "pandas": "pandas", "soundfile": "soundfile", "notebook": "notebook"}
for importnamn, paketnamn in paket.items():
    if importlib.util.find_spec(importnamn) is None:
        print(f"{paketnamn:<11} SAKNAS")
    else:
        try:
            print(f"{paketnamn:<11} OK ({version(paketnamn)})")
        except PackageNotFoundError:
            print(f"{paketnamn:<11} OK")
```

## Git-rutin (se `docs/git_workflow.md`)
1. `git status` före arbetet.
2. Ny vecka: `git switch main`, `git pull origin main`, `git switch -c vecka-NN-kort-beskrivning`.
3. Små commits med specifika filer (inte `git add .`), t.ex. `git commit -m "notebook: ..."`.
4. Fredag: Run All och spara med outputs, skriv `reports/vecka_0N_rapport.md`, pusha och öppna en pull request mot `main`.
